### Dictionaries & Nesting

A dictionary in Python functions similarly to a dictionary in real life. It's a data structure that allows us to associate a key to a value and pair the two pieces of data together.

This is how you create a dicitionary in Python:

````python
# An example dictionary
colours = {
    "apple": "red",
    "pear": "green",
    "banana": "yellow",
}
colours
````

This is how you retrieve items from a dictionary:

````python
colours["pear"]
````

This is how you create an empty dictionary:

````python
my_empty_dictionary = {}
my_empty_dictionary
````

This is how you add new items to an existing dictionary:

````python
colours["peach"] = "pink"
colours
````

This is how you can edit an existing value in a dictionary:

````python
colours["apple"] = "green"
colours["apple"]
````

This is how to loop through a dictionary and print all the keys:

````python
for key in colours:
    print(key)
````

This is how to loop through a dictionary and print all the values:

````python
for key in colours:
    print(colours[key])
````

#### Grading Program

````python
student_scores = {
    'Harry': 88,
    'Ron': 78,
    'Hermione': 95,
    'Draco': 75,
    'Neville': 60
}

student_grades = {}

for k in student_scores:
    
    curr_score = student_scores[k]
    
    if curr_score >= 91:
        student_grades[k] = "Outstanding"
    elif curr_score >= 81:
        student_grades[k] = "Exceeds Expectations"
    elif curr_score >= 71:
        student_grades[k] = "Acceptable"
    else:
        student_grades[k] = "Fail"

student_grades
````

### Nesting

You can mix and match various data types to achieve your desired structure.

#### Nesting a List inside a Dictionary
Instead of a String value assigned to a key, we can replace it with a List. You can nest a List in a Dictionary like this:
```python
my_dictionary = {
    key1: [List],
    key2: Value,
}
```

See if you can figure out how to print out "Lille" from the nested List called travel_log.

````python
travel_log = {
    "France": ["Paris", "Lille", "Dijon"],
    "Germany": ["Stuttgart", "Berlin"],
}
travel_log["France"][1]
````

#### Nesting Lists inside other Lists
We've previously seen Nested Lists:

````python
nested_list = ["A", "B", ["C", "D"]]
````

Do you remember how to get items that are nested deeply in a list? Try to print "D" from the list nested_list.
To get this list: `["C", "D"]` we need the code:

````python
nested_list[2]
````

Therefore, to get "D" we need:

````python
nested_list[2][1]
````

#### Nesting a Dictionary inside a Dictionary
You can also nest a dictionary in a dictionary:

````python
# Figure out how to print out "Stuttgart" from the following list:

travel_log = {
  "France": {
    "cities_visited": ["Paris", "Lille", "Dijon"], 
    "total_visits": 12
   },
  "Germany": {
    "cities_visited": ["Berlin", "Hamburg", "Stuttgart"], 
    "total_visits": 5
   },
}

travel_log["Germany"]["cities_visited"][2]
````
