## **for loop**
```python
for item in list_of_items:
    # Do something to each item
```
Rather than always iterating over an arithmetic progression of numbers (like in Pascal), or giving the user the ability to define both the iteration step and halting condition (as C), Python’s for statement iterates over the items of any sequence (a list or a string), in the order that they appear in the sequence.

````python
# Measure some strings:
words = ["cat", "window", "defenestrate"]
for w in words:
    print(w, len(w), sep=",")
````

If you have data both in list or generator, you can simplify the calculation by using **`sum()`**, **`max()`** and **`min()`** built-in functions.

````python
student_scores = [150, 142, 185, 120, 171, 184, 149, 24, 59, 68, 199, 78, 65, 89, 189, 169, 146]
total_exam_score = sum(student_scores)
total_exam_score
````

````python
# Replicating sum()
s = 0
for score in student_scores:
    s += score
s
````

````python
max_score = max(student_scores)
max_score
````

````python
# Replicating max()
max_score = -1
for score in student_scores:
    if score > max_score:
        max_score = score
max_score
````

### for loops and the **range()** function
```python
for number in range(a, b):
    # Do something with the number
```
If you do need to iterate over a sequence of numbers, the built-in function **`range()`** comes in handy. It generates arithmetic progressions:

````python
n = 100
s = n * (n + 1) // 2
s
````

````python
# range
s = 0
for number in range(1, 101):
    s += number
s
````

````python
# FizzBuzz
to_print = []
for n in range(1, 101):
    if n % 15 == 0:
        to_print.append("FizzBuzz")
    elif n % 3 == 0:
        to_print.append("Fizz")
    elif n % 5 == 0:
        to_print.append("Buzz")
    else:
        to_print.append(n)
to_print[5:15]
````

````python
for i in range(5):
    print(i, end=" ")
````

The given end point is never part of the generated sequence; `range(10` generates 10 values, the legal indices for items of a sequence of length 10. It is possible to let the range start at another number, or to specify a different increment (even negative; sometimes this is called the ‘step’):

````python
list(range(5, 10))
````

````python
list(range(0, 10, 3))
````

````python
list(range(-10, -100, -30))
````

A strange thing happens if you just print a range:

````python
range(10)
````

In many ways the object returned by `range()` behaves as if it is a list, but in fact it isn’t. It is an object which returns the successive items of the desired sequence when you iterate over it, but it doesn’t really make the list, thus saving space.

We say such an object is *iterable*, that is, suitable as a target for functions and constructs that expect something from which they can obtain successive items until the supply is exhausted. We have seen that the *for* statement is such a construct, while an example of a function that takes an iterable is `sum()`:

````python
sum(range(4)) # 0 + 1 + 2 + 3
````

### enumerate function
**`enumerate(iterable, start=0)`**

````python
seasons = ['Spring', 'Summer', 'Fall', 'Winter']
list(enumerate(seasons))
````

````python
list(enumerate(seasons, start=1))
````

````python
# Equivalent to:
def enumerate(iterable, start=0):
    n = start
    for elem in iterable:
        yield n, elem # When you call a function with yield keyword(s), the return value will be a list of values, one for each yield.
        n += 1
````

````python
for i, season in enumerate(seasons):
    print(i, season, end="; ")
````

````python
for tup in enumerate(seasons):
    print(tup, end="; ")
````

````python
seasons_start_month = ["March", "June", "September", "December", "Foo", "Bar", "Baz"]
for s, m in zip(seasons_start_month, seasons):
    print(s, m, sep=", ", end="; ")
````

````python
# Replicating zip()
def zippy(x, y):
    iter1 = iter(x)
    iter2 = iter(y)
    for _ in range(min(len(x), len(y))):
        yield next(iter1), next(iter2)

for s, m in zippy(seasons_start_month, seasons):
    print(s, m, sep=", ", end="; ")
````

### break and continue Statements
The **break** statement breaks out of the innermost enclosing for or while loop:

````python
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0:
            print(f"{n} equals {x} * {n//x}")
            break
````

The **continue** statement continues with the next iteration of the loop:

````python
for num in range(2, 5):
    if num % 2 == 0:
        print(f"Found an even number {num}")
        continue
    print(f"Found an odd number {num}")
````

### else Clauses on Loops

In a *for* or *while* loop the break statement may be paired with an else clause. If the loop finishes without executing the break, the else clause executes.

In a *for* loop, the else clause is executed after the loop finishes its final iteration, that is, if no break occurred.

In a *while* loop, it’s executed after the loop’s condition becomes false.

In either kind of loop, the else clause is **not** executed if the loop was terminated by a break. Of course, other ways of ending the loop early, such as a *return* or a raised exception, will also skip execution of the **else** clause.

````python
for n in range(2, 8):
    for x in range(2, n):
        if n % x == 0:
            print(n, 'equals', x, '*', n//x)
            break
    else:
        # loop fell through without finding a factor
        print(n, 'is a prime number')
````

(Yes, this is the correct code. Look closely: the `else` clause belongs to the `for` loop, not the `if` statement.)

One way to think of the else clause is to imagine it paired with the `if` inside the loop. As the loop executes, it will run a sequence like if/if/if/else. The `if` is inside the loop, encountered a number of times. If the condition is ever true, a `break` will happen. If the condition is never true, the `else` clause outside the loop will execute.

When used with a loop, the `else` clause has more in common with the `else` clause of a `try` statement than it does with that of `if` statements: a `try` statement’s `else` clause runs when no exception occurs, and a loop’s `else` clause runs when no `break` occurs. For more on the try statement and exceptions, see [Handling Exceptions](https://docs.python.org/3/tutorial/errors.html#tut-handling).

### pass Statements

The `pass` statement does nothing. It can be used when a statement is required syntactically but the program requires no action. For example:

````python
# while True:
#     pass  # Busy-wait for keyboard interrupt (Ctrl+C)
````

````python
# This is commonly used for creating minimal classes:
class MyEmptyClass:
    pass
````

Another place `pass` can be used is as a place-holder for a function or conditional body when you are working on new code, allowing you to keep thinking at a more abstract level. The pass is silently ignored:
