#### In Python Data Types are Flexible
So you can create a variable and change it's data type later on. This is known as Dynamic Typing.

````python
age: int
````

````python
age = 22
````

````python
age = "twenty two" # interpreter will try to tell you that this is not expected
````

````python
def police_check(age: int):
    if age > 18:
        can_drive = True
    else: 
        can_drive = False
    return can_drive

print(police_check(19))
````

````python
if police_check(19):
    print("You may pass")
else:
    print("Pay a fine.")
````

````python
# In addition you can also specify the expected output of the function:
def police_check(age: int) -> bool:
    pass
````

The above are known as Type Hints in Python
