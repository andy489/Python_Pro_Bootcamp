## Advanced Python Decorator Functions

````python
## Advanced Python Decorator Functions

class User:
    def __init__(self, name):
        self.name = name
        self.is_logged_in = False

def is_authenticated_decorator(function):
    def wrapper(*args, **kwargs):
        function(args[0]) if args[0].is_logged_in else print(f"User is not logged! Please login first.")
    return wrapper

@is_authenticated_decorator
def create_blog_post(user):
    print(f"This is {user.name.capitalize()}'s new blog post.")
````

````python
pesho = User("pesho")
pesho.is_logged_in = False
create_blog_post(pesho)
````

````python
andrey = User("andrey")
andrey.is_logged_in = True
create_blog_post(andrey)
````

````python
def logging_decorator(function):
    def wrapper(*args, **kwargs):
        print(f"You called {function.__name__}{args}")
        result = function(*args)
        print(f"It returned: {result}")
        return result

    return wrapper


@logging_decorator
def a_function(*args):
    return sum(args)


a_function(1, 2, 3)
````
