#### Creating datetime object

```python
date_of_birth = dt.datetime(year=1988, month=12, day=22)
print(date_of_birth)
```

```python
date_of_birth = dt.datetime(year=1988, month=12, day=22, hour=3)
print(date_of_birth)
```
