## Creating Classes

### Class is a Blueprint for creating an object

```python
import json

class Car:
    def __init__(self):
        self.seats = 5

    def enter_race_mode(self):
        self.seats = 2

    def __str__(self):
        obj_dict = {key: value for key, value in self.__dict__.items()
                    if not key.startswith('__') and not callable(value)}
        
        return json.dumps(obj_dict, indent=4)
        
car = Car()
print(car)
```

```python
car.enter_race_mode()
print(car)
```
