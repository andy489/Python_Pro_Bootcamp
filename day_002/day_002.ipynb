### String

Textual data in Python is handled with `str` objects, or [strings](https://docs.python.org/3/library/stdtypes.html#textseq). Strings are immutable sequences of Unicode code points. String literals are written in a variety of ways:
- Single quotes: `'allows embedded "double" quotes'`
- Double quotes: `"allows embedded 'single' quotes"`
- Triple quoted: `'''Three single quotes'''`, `"""Three double quotes"""`

Triple quoted strings may span multiple lines - all associated whitespace will be included in the string literal.

String literals that are part of a single expression and have only whitespace between them will be implicitly converted to a single string literal. That is, `("spam " "eggs") == "spam eggs"`.

Since there is no separate “character” type, indexing a string produces strings of length 1. That is, for a non-empty string s, `s[0] == s[0:1]`.

```python
string = "literal"
string[0] == "literal"[0]
```

```python
string[4] == string[len(string) - 1]
```

```python
string[4] == string[-1]
```

### Boolean operations
These are the Boolean operations, ordered by ascending priority:
|Operation|Result|Notes|
|:-|:-|:-|
|`x or y`|if *x* is true, then *x* else *y*|(1)|
|`x and y`|if *x* is false, then *x*, else *y*|(2)|
|`not x`|if *x* is false, then `True`, else `False`|(3)|
Notes:

1. This is a short-circuit operator, so it only evaluates the second argument if the first one is false.
2. This is a short-circuit operator, so it only evaluates the second argument if the first one is true.
3. `not` has a lower priority than non-`Boolean` operators, so `not a == b` is interpreted as `not (a == b)`, and `a == not b` is a syntax error.

### Type Conversion

`ord(c)`: Given a string representing one Unicode character, return an integer representing the Unicode code point of that character. For example, `ord('a')` returns the integer 97 and `ord('\u2020')` returns 8224. This is the inverse of `chr()`.

```python
# ASCII
ord('a')
```

`chr(i)`: Return the string representing a character whose Unicode code point is the integer *i*. For example, `chr(97)` returns the string `'a'`. This is the inverse of `ord()`. The valid range for the argument is from 0 through 1,114,111 (0x10FFFF in base 16). `ValueError` will be raised if *i* is outside that range.

```python
chr(97)
```

```python
# int(x, base=10): Return an integer object constructed from a number or string x, or return 0 if no arguments are given.
# str(object=b'', encoding='utf-8', errors='strict'): Return a str version of object

var = "489"
print(var, type(var))
var = int(var)
print(var, type(var))
```

### f-Strings and String Formatting
format(*args, **kwargs): Perform a string formatting operation. The string on which this method is called can contain literal text or replacement fields delimited by braces {}. Each replacement field contains either the numeric index of a positional argument, or the name of a keyword argument. Returns a copy of the string where each replacement field is replaced with the string value of the corresponding argument.

```python
score = 10
height = 1.8
is_winning = True
f"Your score is = {score}, your height is {height}, You are winning is {is_winning}."
```

```python
'{}, {}, {}'.format('a', 'b', 'c')
```

```python
'{0}, {2}, {0}'.format('a', 'b', 'c') # arguments' indices can be repeated
```

```python
'{2}, {1}, {0}'.format(*'abc') # unpacking argument sequence
```

```python
# Accessing arguments by name:
'Coordinates: {latitude}N, {longitude}E'.format(latitude=42.6977, longitude=-23.3219)
```

```python
# Accessing arguments' items:
coord = (3, 5)
'X: {0[0]};  Y: {0[1]}'.format(coord)
```

```python
coord = {'latitude': 42.6977, 'longitude': -23.3219}
'Coordinates: {latitude}N, {longitude}E'.format(**coord)
```

```python
c = 3-5j
('The complex number {0} is formed from the real part {0.real} '
 'and the imaginary part {0.imag}.').format(c)
```

```python
# Accessing arguments’ attributes:
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y
    def __str__(self):
        return 'Point({self.x}, {self.y})'.format(self=self)

str(Point(4, 2))
```

```python
# Replacing %s and %r:
"repr() shows quotes: {!r}; str() doesn't: {!s}".format('test1', 'test2')
```

```python
# Aligning the text and specifying a width:
'{:<30}'.format('left aligned')
```

```python
'{:^30}'.format('centered')
```

```python
'{:*^30}'.format('centered')  # use '*' as a fill char
```

```python
# Replacing %+f, %-f, and % f and specifying a sign:
'{:+f}; {:+f}'.format(3.14, -3.14) # show it always
```

```python
'{: f}; {: f}'.format(3.14, -3.14) # show a space for positive numbers
```

```python
'{:-f}; {:-f}'.format(3.14, -3.14) # show only the minus -- same as '{:f}; {:f}'
```

```python
# Replacing %x and %o and converting the value to different bases:
"int: {0:d};  hex: {0:x};  oct: {0:o};  bin: {0:b}".format(42) # format also supports binary numbers
```

```python
"int: {0:d};  hex: {0:#x};  oct: {0:#o};  bin: {0:#b}".format(42) # with 0x, 0o, or 0b as prefix:
```

```python
# Using the comma as a thousands separator:
'{:,}'.format(1234567890)
```

```python
# Expressing a percentage:
points = 19
total = 22
'Correct answers: {:.2%}'.format(points/total)
```

```python
# Using type-specific formatting:
import datetime
d = datetime.datetime(2010, 7, 4, 12, 15, 58)
'{:%Y-%m-%d %H:%M:%S}'.format(d)
```

```python
# Nesting arguments and more complex examples:
width = 5
for num in range(5,12):
    for base in 'dXob':
        print('{0:{width}{base}}'.format(num, base=base, width=width), end=' ')
    print()
```

```python
for align, text in zip('<^>', ['left', 'center', 'right']):
    print('{0:{fill}{align}16}'.format(text, fill=align, align=align))
```

```python
# \r moves the cursor to the start of the line, and \n moves it to the next line

# When printing to the console, \r can be used to update the same line repeatedly,
# which is useful for progress bars or dynamic updates.

import time
for i in range(21):
    print(f"\r[{'#' * i}{' ' * (20 - i)}] {i * 5}%", end="", flush=True)
    time.sleep(0.1)
print()
```
