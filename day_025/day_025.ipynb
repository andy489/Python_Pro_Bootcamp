## CSV Data and the Pandas Library

### Read csv data file with csv library

```python
import csv

with open("data/weather_data.csv") as data_file:
    data = csv.reader(data_file)

    temperatures = []

    for row in data:
        if row[1] != "temp":
            temperatures.append(int(row[1]))

    print(temperatures)
```

#### The Great Squirrel Census Data Analysis
#### https://data.cityofnewyork.us/Environment/2018-Central-Park-Squirrel-Census-Squirrel-Data/vfnx-vebw/about_data

```python
squirrel_data = pd.read_csv("data/2018_Central_Park_Squirrel_Data.csv")

squirrel_data["Primary Fur Color"]
```

```python
grey_squirrels = squirrel_data[squirrel_data["Primary Fur Color"] == "Gray"]
grey_squirrels
```

```python
grey_squirrels_count = len(grey_squirrels)
red_squirrels_count = len(squirrel_data[squirrel_data["Primary Fur Color"] == "Cinnamon"])
black_squirrels_count = len(squirrel_data[squirrel_data["Primary Fur Color"] == "Black"])
grey_squirrels_count, red_squirrels_count, black_squirrels_count
```

```python
data_dict = {
    "Fur Color": ["Gray", "Cinnamon", "Black"],
    "Count": [grey_squirrels_count, red_squirrels_count, black_squirrels_count],
}
data_dict
```

```python
df = pd.DataFrame(data_dict)
# df.to_csv("data/squirrels_count.csv")
df
```
