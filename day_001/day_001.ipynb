### Fancier Output Formatting

```python
# Basic usage of the str.format() method 
print("1. Mix {1}g of Flour, {2}g Yeast and {0}ml Water in a bowl.".format(10, 100, 500))

# Keyword arguments are referred to by using the name of the argument.
print("2. {action} the dough for {mins} minutes.".format(mins=10, action="Knead"))

# By passing the table dictionary as keyword arguments with the ** notation
print("3. Add {salt}g of Salt.".format(**{'salt':3}))

# Old string formatting
print(f"4. Leave to rise for %.0f hours." % 2)

# Using f-string
degrees = 200
minutes = 30
print(f"5. Bake at {degrees} degrees C for {minutes} minutes.")
```

### Special Characters
- `\` special `escape` character. Used to insert characters that are illegal in a string.
- `\n` for *newline*. Use `\n` when working in environments where the platform is known and controlled. Opting for os.linesep in scripts and applications designed for cross-platform use.
- `\r` for carriage return**. Moves the cursor to the beginning of the current line without advancing to the next line.
- `\t` for `tab`. It is interpreted by the editor as an amount of white space, usually 4 or 8 spaces.

```python
print("Some illegal chars: \\ (backslash), \" (double quotes inside double-quoted strings)")
```

```python
print("Hello,\nPython!")
```

```python
print("Hello,\rPython!")
```

```python
print("Hello,\tPython!")
```

### Swap Two Variables in Python

```python
glass1 = "milk"
glass2 = "juice"
print(glass1, glass2)

# Using a third temporary variable
tmp = glass1
glass1 = glass2
glass2 = tmp

print(glass1, glass2)

# Or do it the Python way
glass1, glass2 = glass2, glass1

print(glass1, glass2)
```

### Naming Variables
#### Rules and Conventions for Python Variables:

- A variable name must start with a letter or the underscore character;
- A variable name can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ );
- Variable names are case-sensitive (age, Age and AGE are three different variables);
- Use **snake_case** naming convention when there is a variable with a collection of words;
- A variable name cannot be any of the Python keywords.
