### Control Flow with if/else and Conditional Operators

#### Comparisons (Relations)

There are eight comparison operations in Python. They all have the same priority (which is higher than that of the Boolean operations). Comparisons can be chained arbitrarily; for example, `x < y <= z` is equivalent to `x < y` and `y <= z`, except that `y` is evaluated only once (but in both cases `z` is not evaluated at all when `x < y` is found to be false).

This table summarizes the comparison operations:

|Operation|Meaning|
|:-|:-|
|`<`|strictly less than|
|`<=`|less than or equal|
|`>`|strictly greater than|
|`>=`|greater than or equal|
|`==`|equal|
|`!=`|not equal|
|`is`|bject identity|
|`is not`|negated object identity|

#### Truth Value Testing
Any object can be tested for truth value, for use in an *if* or *while* condition or as operand of the Boolean operations `or`, `and` and `not`.

By default, an object is considered true unless its class defines either a `__bool__()` method that returns `False` or a `__len__()` method that returns zero, when called with the object. Here are most of the built-in objects considered false:

- constants defined to be false: `None` and `False`
- zero of any numeric type: `0`, `0.0`, `0j`, `Decimal(0)`, `Fraction(0, 1)`
- empty sequences and collections: `''`, `()`, `[]`, `{}`, `set()`, `range(0)`

#### Check Odd or Even

```python
n = int(input("Enter a whole number: "))
if n & 1 == 0:
    print(f"{n} is even")
else: 
    print(f"{n} is odd")
```

#### BMI calculator with interpretations

```python
weight = 78
height = 1.77

bmi = round(weight / height ** 2, 1)

if bmi < 18.5:
    print("underweight")
elif bmi < 25:
    print("normal weight")
else:
    print("overweight")
```

#### Rollercoaster cash register

```python
print("Welcome to the rollercoaster!")
height = int(input("What is your height in cm? "))
bill = 0
if height > 120:
    print("You can ride the rollercoaster")
    age = int(input("What is your age? "))
    if age <= 12:
        bill += 5
        print("Child tickets are $5.")
    elif age <= 18:
        bill += 7
        print("Youth tickets are $7.")
    elif 45 <= age <= 55: # age >= 45 and age <= 55
        print("Everything is going to be OK. Have a free ride on us")
    else:
        bill += 12
        print("Adult tickets are $12.")
    wants_photo = input("Do you want to have a photo taken? Type 'y' for Yes and 'n' for No.").lower()
    if wants_photo == "y":
        bill += 3
    print(f"Your final bill is ${bill}")
else:
    print("Sorry you have to grow taller before you can ride.")
```

#### Pizza order

```python
print("Welcome to Python Pizza Deliveries!")

size = input("What size pizza do you want? S, M or L: ").upper()
pepperoni = input("Do you want popperoni on your pizza? Y or N: ").upper()
extra_cheese = input("Do you want extra cheese? Y or N: ").upper()

bill = 0

if size == "S":
    bill += 15
elif size == "M":
    bill += 20
else:
    bill += 25
    
if pepperoni == "Y":
    if size == "S":
        bill += 2
    else:
        bill += 3
        
if extra_cheese == "Y":
    bill += 1
    
print(f"Your final bill is: ${bill}.")
```

### Match-Case
Starting from Python 3.10, the Python language has introduced the match statement

```python
# data = 1       # Output: Data is 1.
# data = "hello" # Output: Data is 'hello'.
data = [4, 8]  # Output" Data is a list with elements 4 and 8.
# data = 2       # Output: Data is a positive number: 2.
# data = {}      # Output: Data does not match any pattern.

match data:
    case 1:
        print("Data is 1.")
    case "hello":
        print("Data is 'hello'.")
    case [x, y]:
        print(f"Data is a list with elements {x} and {y}.")
    case x if type(x) is type(0) and x > 0:
        print(f"Data is a positive number: {x}.")
    case _:
        print("Data does not match any pattern.")
```

```python
# data = {"type": "json", "content": "{\"match-case\": \"is awsome\"}"} # Output: JSON data. Content: {"match-case": "is awsome"}
data = {"type": "xml", 
        "content": "<match_case><is>awesome</is></match_case>"} # Output: XML data. Content: <match_case><is>awesome</is></match_case>
# data = {"type": "csv", "content": "match-case,is,awesome"} # Output: CSV data. Content: match-case,is,awesome

match data:
    case {"type": "json", "content": content}:
        print(f"JSON data. Content: {content}")
    case {"type": "xml", "content": content}:
        print(f"XML data. Content: {content}")
    case {"type": "csv", "content": content}:
        print(f"CSV data. Content: {content}")
    case _:
        raise ValueError("Unsupported data format.")
```

### Logical Operators

```python
# De Morgan's Laws
A = True
B = False
print(not (A and B) is not A or not B)
print(not(A or B) is not A and not B)
```
