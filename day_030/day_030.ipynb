## Errors, Exceptions and JSON Data

#### Handling Errors and Exceptions

```python
# FileNotFound
try:
    file = open("no_such_file.txt")
except FileNotFoundError as ex:
    print(ex)
```

```python
# KeyError
a_dict = {"key": "value"}
try:
    value = a_dict["no_such_key"]
except KeyError as ex:
    print(ex)
```

```python
# IndexError
fruit_list = ["Apple", "Banana", "Pear"]
try:
    take_fruit = fruit_list[3]
except IndexError as ex:
    print(ex)
```

```python
# TypeError
text = "abc"
try:
    print(text + 5)
except TypeError as ex:
    print(ex)
```

```python
try:
    file = open("data/a_file.txt")
    a_dict = {"key": "value"}
    print(a_dict["key"])
except FileNotFoundError:
    file = open("data/a_file.txt", "w")
    file.write("something in file")
except KeyError as erro_msg:
    print(f"The key {erro_msg} does not exist")
else:
    content = file.read()
    print(content)
finally:
    file.close()
    print("File was closed.")
```

#### Raising your own Exceptions

```python
#BMI Example

height = float(input("Height: "))
weight = int(input("Weight: "))

if height > 3:
    raise ValueError("Human Height should not be over 3 meters.")

bmi = weight / height ** 2
print(bmi)
```

```python
fruits = ["Apple", "Pear", "Orange"]
 
# Catch the exception and make sure the code runs without crashing.
def make_pie(index):
  try:
    fruit = fruits[index]
  except IndexError:
    print("Fruit pie")
  else:
    print(fruit + " pie")
 
make_pie(4)
```

```python
facebook_posts = [
    {'Likes': 21, 'Comments': 2},
    {'Likes': 13, 'Comments': 2, 'Shares': 1},
    {'Likes': 33, 'Comments': 8, 'Shares': 3},
    {'Comments': 4, 'Shares': 2},
    {'Comments': 1, 'Shares': 1},
    {'Likes': 19, 'Comments': 3},
] 
 
def count_likes(posts):
    total_likes = 0
    for post in posts:
      try:
        total_likes = total_likes + post['Likes']
      except KeyError:
        pass 
    
    return total_likes
 
 
count_likes(facebook_posts)
```

#### Write, read and update JSON data

Python JSON Module Documentation: https://docs.python.org/3/library/json.html

```python
# Custom Serialization
import json
def custom_json(obj):
    if isinstance(obj, complex):
        return {'__complex__': True, 'real': obj.real, 'imag': obj.imag}
    raise TypeError(f'Cannot serialize object of {type(obj)}')

json.dumps(1 + 2j, default=custom_json)
```
