```python
# https://reeborg.ca/reeborg.html?name=problem_world3.json&url=user_world%3Aproblem_world3.json
  
def turn_left_n(n):
    for _ in range(n):
        turn_left()

def turn_right():
    turn_left_n(3)

def maze_escape_solution_1():
    """Detects and eleminates cycles from the robot's path.
    Not very efficient for large rectangular free spaces without walls along the maze"""
    right_turn_cnt = 0
    while not at_goal():
        if right_is_clear() and right_turn_cnt < 3:
            right_turn_cnt += 1
            turn_right()
            move()
        elif front_is_clear():
            right_turn_cnt = 0
            move()
        else:
            right_turn_cnt = 0
            turn_left()

def maze_escape2():
    """Robot traverses forward until it finds a wall."""
    while front_is_clear():
        move()
    turn_left()    
    
    while not at_goal():
        if right_is_clear():
            turn_right()
            move()
        elif front_is_clear():
            move()
        else:
            turn_left()
        
maze_escape2()
```
